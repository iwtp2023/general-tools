# Code to Generate Triplicate Line Graphs
The following code is useful for generating line graphs of triplicate runs, for example of MD RMSD or ```thermal_mmgbsa.py``` analysis.

````python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
````

````python
# -- font and color formatting options --
mpl.rc('font', **{'family': 'sans-serif', 'sans-serif': ['Arial'], 'size': 15})
palette = ['mediumturquoise', 'mediumseagreen', 'darkcyan']
````

````python
# load RMSD data in csv format
data = pd.read_csv('./epicatechin_MD.csv', header=0)
````

````python
plt.figure(figsize=(8, 5))

line1, = plt.plot(data["Time (ns)"], data["run_1"], label="Run 1", color=palette[0])
line2, = plt.plot(data["Time (ns)"], data["run_2"], label="Run 2", color=palette[1])
line3, = plt.plot(data["Time (ns)"], data["run_3"], label="Run 3", color=palette[2])

plt.xlabel("Time (ns)")
plt.ylabel("Epicatechin RMSD (Å)")

plt.ylim(-1, 9) # adjust appropriately for the data

legend = plt.legend(loc="upper right",handlelength=0, frameon=False, prop={"weight": "bold"})
for text, line in zip(legend.get_texts(), [line1, line2, line3]):
    text.set_color(line.get_color())

plt.subplots_adjust(bottom=0.2)
plt.margins(x=0)
plt.grid(False)
plt.show()
````

## Probability Density Plots
When you would like to see or compare the distribution of your data points, rather than a time series, it is useful to generate a probability density plot. A probability density fits a smooth function to the histogram of the data. 

````python
data = pd.read_csv('./metal_coordination_dist.csv', header=0)
````

````python
colors = ["darkcyan", "limegreen", "mediumorchid"]
````

````python
fig, ax = plt.subplots(figsize=(8, 4.4))
for col, color in zip(data.columns, colors):
    sns.kdeplot(data[col], ax=ax, color=color, label=col)

legend = ax.legend(frameon=False, prop={"weight": "bold"})
legend.set_title(None)

ax.set_xlabel('Ligand-Copper Coordination Distance (Å)')
ax.set_ylabel('Density')

# optional - removes lines before the legend labels
for handle in legend.legend_handles:
    handle.set_visible(False)

# matches legend text color to line color - always use if handles are invisible
for text, line in zip(legend.get_texts(), ax.lines):
    text.set_color(line.get_color())

plt.show()
````
